## Section S03 — Query Analysis & OptimizationAnalyze, optimize, and execute queries using the production RAG pipeline.

```python
from aifluent.query.analyzer import QueryAnalyzerfrom aifluent.query.optimizer import QueryOptimizeranalyzer = QueryAnalyzer()optimizer = QueryOptimizer()query = "What is the revenue impact of the new product launch in Q3?"analysis = analyzer.analyze(query)print(f'Query:       {analysis.original}')print(f'Intent:      {analysis.intent}')print(f'Keywords:    {analysis.keywords}')print(f'Entities:    {analysis.entities}')print(f'Is question: {analysis.is_question}')
```

## Query Optimization Strategies

```python
for strategy in ["rewrite", "expand", "remove_stopwords", "all"]:    expanded = optimizer.optimize(query, strategy=strategy)    print(f'{strategy}: {len(expanded)} variants')    for v in expanded[:3]:        print(f'  - {v[:80]}')
```
